```python
import os
from utils import get_openai_api_key

openai_api_key = get_openai_api_key()
os.environ["OPENAI_MODEL_NAME"] = 'gpt-6.1-sol'
```

```python

planner = Agent(
    role="Educational Content Planner",

    goal=(
        "Plan engaging, informative, and factually accurate "
        "educational blog articles on {topic} for students "
        "of classes 6 to 10 at Ramanujan Classes. "
        "Create a clear article structure that helps students "
        "understand the topic and provides useful information "
        "to parents."
    ),

    backstory=(
        "You are an experienced educational content planner "
        "working for Ramanujan Classes, an educational institute "
        "for students of classes 6 to 10. "
        "You understand school-level mathematics, science, "
        "English, and general academic development. "
        "You research educational topics, identify important "
        "points, organize information into logical sections, "
        "and create engaging article outlines. "
        "Your plans help the Content Writer produce accurate, "
        "student-friendly, and informative blog articles."
    ),

    allow_delegation=False,
    verbose=True
)
```

```python

writer = Agent(
    role="Educational Content Writer",

    goal=(
        "Write engaging, informative, and factually accurate "
        "educational blog articles about {topic} for students "
        "of classes 6 to 10 at Ramanujan Classes. "
        "Create easy-to-understand content with clear "
        "explanations, practical examples, and useful "
        "learning tips."
    ),

    backstory=(
        "You are an experienced educational content writer "
        "working for Ramanujan Classes. "
        "You specialize in writing educational articles "
        "for school students from classes 6 to 10. "
        "You use the research and article outline provided "
        "by the Content Planner to write complete articles. "
        "You explain difficult concepts in simple English, "
        "include practical examples, and organize content "
        "with meaningful headings and subheadings. "
        "You ensure that the article is factually accurate, "
        "student-friendly, informative, and easy to understand. "
        "Avoid unsupported claims, unnecessary repetition, "
        "and overly complicated language."
    ),

    allow_delegation=False,
    verbose=True
)
```

```python

editor = Agent(
    role="Educational Content Editor",

    goal=(
        "Review and edit educational blog articles about {topic} "
        "for Ramanujan Classes. Ensure that the article is "
        "factually accurate, grammatically correct, well-structured, "
        "easy to understand, and suitable for students "
        "of classes 6 to 10."
    ),

    backstory=(
        "You are an experienced educational content editor "
        "working for Ramanujan Classes. "
        "You receive educational articles from the Content Writer "
        "and review them before publication. "
        "You check grammar, spelling, sentence structure, "
        "factual accuracy, clarity, and consistency. "
        "You ensure that explanations are appropriate for "
        "school students from classes 6 to 10. "
        "You improve headings, paragraphs, and transitions "
        "to make articles engaging and easy to read. "
        "You remove repetitive content, unsupported claims, "
        "and unnecessary information. "
        "You preserve the original meaning while improving "
        "the overall quality of the article. "
        "You ensure that the final article maintains a "
        "professional and educational tone suitable for "
        "Ramanujan Classes."
    ),

    allow_delegation=False,
    verbose=True
)
```

```python

plan = Task(
    description=(
        "Create a detailed educational blog article plan "
        "for Ramanujan Classes on the topic: {topic}.\n\n"

        "1. Research and identify the important concepts "
        "and key information related to {topic}.\n"

        "2. Identify the target audience, focusing on "
        "students of classes 6 to 10 and their parents.\n"

        "3. Understand the learning needs and common "
        "difficulties students may face regarding this topic.\n"

        "4. Develop a detailed article outline containing "
        "an introduction, main headings, subheadings, "
        "examples, and a conclusion.\n"

        "5. Suggest practical examples and learning tips "
        "that students can understand easily.\n"

        "6. Include relevant SEO keywords and suggest "
        "an engaging article title.\n"

        "7. Ensure that the content plan is educational, "
        "informative, and appropriate for Ramanujan Classes."
    ),

    expected_output=(
        "A comprehensive educational content plan containing "
        "an article title, target audience analysis, "
        "introduction, structured headings and subheadings, "
        "key learning points, practical examples, "
        "conclusion, and relevant SEO keywords."
    ),

    agent=planner,
)
```

```python

write = Task(
    description=(
        "1. Use the educational content plan provided by "
        "the Content Planner to write a complete blog article "
        "on {topic} for Ramanujan Classes.\n"

        "2. Write specifically for students of classes 6 to 10 "
        "and their parents.\n"

        "3. Incorporate relevant SEO keywords naturally "
        "without keyword stuffing.\n"

        "4. Create engaging and meaningful section headings "
        "and subheadings.\n"

        "5. Structure the article with an interesting "
        "introduction, informative main body, practical "
        "examples, useful learning tips, and a conclusion.\n"

        "6. Use simple English so that school students "
        "can easily understand the content.\n"

        "7. Ensure that educational explanations are "
        "factually accurate and avoid unsupported claims.\n"

        "8. Maintain a professional, informative, and "
        "student-friendly tone suitable for Ramanujan Classes."
    ),

    expected_output=(
        "A complete educational blog article in Markdown "
        "format, ready for review and publication on the "
        "Ramanujan Classes website. Include an engaging title, "
        "introduction, properly structured headings, "
        "informative paragraphs, practical examples, "
        "learning tips, conclusion, and naturally integrated "
        "SEO keywords. Each major section should contain "
        "2 to 3 meaningful paragraphs."
    ),

    agent=writer,
)
```

```python

edit = Task(
    description=(
        "Review and proofread the educational blog article "
        "written by the Content Writer about {topic} "
        "for Ramanujan Classes.\n\n"

        "1. Correct grammatical, spelling, and punctuation errors.\n"

        "2. Check whether the article is easy to understand "
        "for students of classes 6 to 10.\n"

        "3. Ensure that the article maintains a professional, "
        "educational, and student-friendly tone.\n"

        "4. Improve sentence structure, headings, "
        "paragraph flow, and readability.\n"

        "5. Remove unnecessary repetition and irrelevant content.\n"

        "6. Identify unsupported claims or potentially "
        "incorrect educational information and flag them "
        "for verification rather than inventing corrections.\n"

        "7. Ensure that SEO keywords are used naturally.\n"

        "8. Preserve the original meaning and important "
        "information while improving the article's quality.\n"

        "9. Prepare the final article for publication "
        "on the Ramanujan Classes website."
    ),

    expected_output=(
        "A professionally edited educational blog article "
        "in Markdown format, ready for publication on "
        "the Ramanujan Classes website. Include a clear title, "
        "well-organized headings, grammatically correct "
        "paragraphs, student-friendly explanations, and "
        "a conclusion. Flag any claims requiring factual "
        "verification."
    ),

    agent=editor
)
```

```python
crew = Crew(
    agents=[planner, writer, editor],
    tasks=[plan, write, edit],
    verbose=True
)
```

```python
import os

# Set the environment variable so CrewAI can find it
os.environ["OPENAI_API_KEY"] = openai_api_key
os.environ["OPENAI_MODEL_NAME"] = 'gpt-3.5-turbo'

result = await crew.kickoff_async(
    inputs={
        "topic": "Importance of Mathematics in Everyday Life"
    }
)

print(result)
```
